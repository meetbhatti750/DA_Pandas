```python
import pandas as pd
```

```python
data = {
    "song": ["Song A", "Song B", "Song A", "Song C"],
    "artist": ["Artist 1", "Artist 2", "Artist 1", "Artist 3"],
    "streams": [1000, 2000, 1000, 3000],
    "date": ["2026-01-01", "2026-01-02", "2026-01-01", "2026-01-03"]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Data:")
print(df)
```

```python
df = df.drop_duplicates(subset=["song"], keep="first")

print("\nAfter Removing Duplicates:")
print(df)
```
