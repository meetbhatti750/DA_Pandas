```python
import pandas as pd
from scipy.stats import zscore

data = {
    "date": [
        "2026-01-01",
        "2026-01-02",
        "2026-01-03",
        "2026-01-04",
        "2026-01-05",
        "2026-01-06"
    ],
    "tickets_sold": [100, 120, 110, 115, 105, 500]
}

df = pd.DataFrame(data)

df["z_score"] = zscore(df["tickets_sold"])

outliers = df[df["z_score"].abs() > 2]

print("Outlier Dates:")
print(outliers["date"])
```
