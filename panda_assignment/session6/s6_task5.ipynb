```python
import pandas as pd
def find_outliers(df, column="transaction_amount",threshold=2):
    df = df.copy()

    mean_value = df[column].mean()
    std_value = df[column].std(ddof=0)  

    if std_value == 0 or pd.isna(std_value):
        return df.iloc[0:0]  
    df["z_score"] = (df[column] - mean_value) / std_value
    return df[df["z_score"].abs() > threshold]

data = {
    "transaction_id": [1, 2, 3, 4, 5, 6, 7],
    "transaction_amount": [100, 150, 200, 250, 300, 10000, 350]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

result = find_outliers(df)
print("outliers transaction:")
print(result)
```
