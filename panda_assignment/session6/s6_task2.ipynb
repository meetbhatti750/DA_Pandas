```python
import pandas as pd
```

```python
data = {
    "product_id": [101, 102, 101, 103, 102],
    "user_id": [1, 2, 1, 3, 2],
    "rating": [5, 4, 3, 5, 2],
    "review_text": ["Good", "Nice", "Average", "Excellent", "Okay"]
}
```

```python
df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
before = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("\nDuplicates before removal:", before)
```

```python

df = df.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)
```

```python
after = df.duplicated(
    subset=["product_id", "user_id"]
).sum()
```

```python
print("Duplicates after removal:", after)

print("\nCleaned Data:")
print(df)
```
