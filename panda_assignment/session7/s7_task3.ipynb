```python
import pandas as pd
```

```python
data = {
    "PostID": [101, 102, 103, 104],
    "Likes": [1500, 2000000, 850, 45000],
    "Comments": [1200, 2500000, 500, 12000]
}

df = pd.DataFrame(data)

df["Likes"] = df["Likes"].apply(
    lambda x: str(round(x/1000, 1)) + "K" if x >= 1000 else str(x)
)

df["Comments"] = df["Comments"].apply(
    lambda x: str(round(x/1000, 1)) + "K" if x >= 1000 else str(x)
)

print(df)
```
