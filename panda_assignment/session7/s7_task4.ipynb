```python
import pandas as pd
```

```python
data = {
    "Restaurant": ["Pizza", "Burger", "Sandwich", "Pasta"],
    "Price": [200, 100, 80, 150],
    "Qty": [2, 3, 1, 2],
    "DeliveryCharge": [30, 20, 10, 15]
}
```

```python
df = pd.DataFrame(data)

df["FinalAmount"] = df.apply(
    lambda x: (x["Price"] * x["Qty"]) + x["DeliveryCharge"],
    axis=1
)

print(df)
```
