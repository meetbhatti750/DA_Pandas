```python
import pandas as pd
```

```python
categories = pd.Series([
    "Electronics/Mobiles/Smartphones",
    "Home/Kitchen/Appliances",
    "Fashion/Men/Shirts",
    "Beauty/Makeup/Lipstick"
])
```

```python

df = pd.DataFrame({"Category": categories})

df["MainCategory"] = df["Category"].str.split("/").str[0]

print(df)
```
