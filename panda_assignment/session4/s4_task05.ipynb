```python
import pandas as pd
```

```python
products = pd.DataFrame({
    "product_name": ["Mobile", "Laptop", "Headphones",
                     "Smartwatch", "Keyboard"],
    "price": [15000, 45000, 2000, 3000, 1000],
    "discount": [10, 20, 35, 15, 25]
})
```

```python
result = products.sort_values(
    by="discount", ascending=False
).reset_index(drop=True)
```

```python
print(result)
```
