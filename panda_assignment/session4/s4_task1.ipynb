```python
import pandas as pd
```

```python
movies = pd.DataFrame({
    "title": ["Dangal", "3 Idiots", "Bajrangi Bhaijaan",
              "PK", "Chhichhore", "Drishyam"],
    "rating": [8.3, 8.4, 8.1, 8.1, 8.3, 8.2],
    "release_year": [2016, 2009, 2015, 2014, 2019, 2015]
})
```

```python
result = movies.sort_values(by="rating", ascending=False)
```
