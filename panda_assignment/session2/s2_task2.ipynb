```python
import pandas as pd
```

```python
expenses = pd.DataFrame({
    "Date": ["01-10-2026", "02-10-2026", "03-10-2026",
             "04-10-2026", "05-10-2026"],
    "Expense": ["Recharge", "Tea", "Travel", "Food", "Internet"],
    "Amount": [299, 30, 100, 150, 399]
})

expenses.to_excel("mobile_expenses.xlsx", index=False)

print("Excel file created!")
```
