```python
import pandas as pd
```

```python
followers = pd.Series([1200, 1500, 800, 2000])

print("Followers:")
print(followers)

```

```python
print("Highest followers:", followers.max())
print("Index of highest:", followers.idxmax())
```

```python
print("Lowest followers:", followers.min())
print("Index of lowest:", followers.idxmin())
```
