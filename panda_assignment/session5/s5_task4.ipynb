```python
import pandas as pd
import numpy as np
```

```python
songs = pd.DataFrame({
    "song": ["Song A", "Song B", "Song C", "Song D", "Song E"],
    "artist": ["Artist 1", "Artist 2", "Artist 3", "Artist 4", "Artist 5"],
    "duration": [180, 210, np.nan, 240, np.nan]
})
```

```python
median_duration = songs["duration"].median()

songs["duration"] = songs["duration"].fillna(median_duration)

songs.to_csv("cleaned_playlist.csv", index=False)
```
