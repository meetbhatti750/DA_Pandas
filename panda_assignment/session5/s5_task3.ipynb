```python
import pandas as pd
import numpy as np
```

```python
products = pd.DataFrame({
    "product_name": ["Mobile", "Laptop", "Headphones", "Tablet", "Mouse"],
    "rating": [4.5, 4.0, np.nan, 3.5, np.nan]
})
```

```python
mean_rating = products["rating"].mean()

products["rating"] = products["rating"].fillna(mean_rating)
```

```python
print(products)
```
